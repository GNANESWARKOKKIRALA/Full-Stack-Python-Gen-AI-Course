```python
"""1️⃣ Stack Problem – Push and Pop Elements
Problem

Create a stack, add 3 numbers, and remove the top element.

Explanation

Push → add element to stack

Pop → remove last element

Stack follows LIFO (Last In First Out).

Example:
Push → 10, 20, 30
Stack → [10, 20, 30]
Pop → removes 30"""
stack=[]
stack.append(10)
stack.append(20)
stack.append(30)
print(stack)
stack.pop(-1)
print(stack)
print(stack[-1])
```

```python
#Stack = LIFO (Last In First Out)
class stack:
    def __init__(self):
        self.stack=[]
    def push(self,data):
        self.stack.append(data)
        print(data)
    def pop(self):
        return "pop value:",self.stack.pop()
    def peek(self):
        return "peek value:",self.stack[-1]
s=stack()
s.push(10) 
s.push(20)
s.push(30) 
print()
print(s.peek())
print(s.pop())
```

```python
"""3️⃣ Queue Problem – Enqueue and Dequeue
Problem

Create a queue, add 3 numbers, then remove the first element.

Explanation

Queue follows FIFO (First In First Out).

Example
Add → 10, 20, 30
Queue → [10, 20, 30]

Dequeue → removes 10"""
from collections import deque
queue=deque()
queue.append(10)
queue.append(20)
queue.append(30)
print(queue)
queue.popleft()
print(queue)
print(queue[0])
```

```python
#QUEUE BASIC CODE (FIFO)
class queue:
    def __init__(self):
        self.queue=[]
    def enqueue(self,data):
        self.queue.append(data)
        print(data)
    def dequeue(self):
        return "dequeue value:",self.queue.pop(0)
    def front(self):
        return "front value:",self.queue[0]
q=queue()
q.enqueue(10)
q.enqueue(20)
q.enqueue(30)
print(q.dequeue())
print(q.front())
```

```python
"""1️⃣ Stack Problem – Reverse a String
Problem

Reverse a string using a stack.

Example
Input: "GYM"
Output: "MYG"

Explanation

Stack follows LIFO.
So the last character inserted will come out first, which reverses the string.

Steps:

Push all characters into stack

Pop characters one by one

Join them"""
stack = []
string = "GYM"

# push characters
for ch in string:
    stack.append(ch)

reversed_string = ""

# pop characters
while stack:
    reversed_string += stack.pop()

print("Reversed string:", reversed_string)
```

```python
"""📌 Problem Statement

You are given a queue of numbers.

👉 Check whether a given number is present in the queue or not.

🧾 Example

Queue = [10, 20, 30, 40]                      
Search = 30

Output:
👉 Found"""
from collections import deque
queue=deque([10,20,30,40,50])
search=40
found=False
for i in queue:
    if i==search:
        found=True
        break
if found:
    print("Found")
else:
    print("not found")
```

```python
#Insert at beginning and End
# Linkedlist
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

# Creating nodes
n1 = Node(10)
n2 = Node(20)
n3 = Node(30)

# Linking nodes
n1.next = n2
n2.next = n3

# 👉 Insert at beginning
new_node = Node(5)
new_node.next = n1
n1 = new_node   # update head

# 👉 Insert at end
temp = n1
while temp.next:
    temp = temp.next
temp.next = Node(40)

# Printing linked list
temp = n1
while temp:
    print(temp.data, end=" -> ")
    temp = temp.next
```
