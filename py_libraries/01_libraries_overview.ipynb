```python
from sklearn.linear_model import LinearRegression

X = [[1], [2], [3]]
y = [2, 4, 6]

model = LinearRegression()
model.fit(X, y)

print(model.predict([[4]]))
```

```python
import xgboost as xgb

X = [[1,2],[2,3],[3,4]]
y = [0, 1, 0]

model = xgb.XGBClassifier()
model.fit(X, y)

print(model.predict([[2,2]]))
```

```python
import sys

print("🔍 Python Executable:")
print(sys.executable)
print("\n🔍 Testing Installed Libraries...\n")

def test_lib(name, import_name=None):
    try:
        module = __import__(import_name or name)
        version = getattr(module, "__version__", "unknown")
        print(f"✅ {name} OK (version: {version})")
    except Exception as e:
        print(f"❌ {name} FAIL → {e}")

libs = [
    ("numpy", None),
    ("pandas", None),
    ("scipy", None),
    ("sklearn", "sklearn"),
    ("xgboost", None),

    ("torch", None),
    ("torchvision", None),
    ("torchaudio", None),

    ("transformers", None),
    ("sentence_transformers", "sentence_transformers"),
    ("openai", None),

    ("langchain", None),
    ("langchain_openai", "langchain_openai"),

    ("sqlalchemy", None),
    ("pymysql", None),
    ("psycopg2", None),

    ("requests", None),
    ("dotenv", "dotenv"),
    ("tqdm", None),
]

for name, imp in libs:
    test_lib(name, imp)

print("\n✔ DONE TESTING ENVIRONMENT")
```
